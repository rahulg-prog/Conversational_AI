```python
from docling.document_converter import DocumentConverter

source = "https://www.honda2wheelersindia.com/services/maintenance/extended-warranty"  # file path or URL
converter = DocumentConverter()
doc = converter.convert(source).document

print(doc.export_to_markdown())  # output: "### Docling Technical Report[...]"
```

```python
# ...existing code...
markdown_content = doc.export_to_markdown()

# Define the file path (e.g., saving to the scraped_markdown folder)
file_path = "../../scraped_markdown/honda_extended_warranty.md"

with open(file_path, "w", encoding="utf-8") as f:
    f.write(markdown_content)

print(f"File saved to {file_path}")
```

```python
import os
from langchain_core.documents import Document

def iterate_files(folder_path):
    docs = []
    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)
        if os.path.isfile(file_path):
            with open(file_path, "r", encoding="utf-8") as f:
                content = f.read()
                doc = Document(page_content=content, metadata={"source": file_name})
                docs.append(doc)
    return docs

# Example: Read the file content and create a Document object
folder_path = r"C:\github_work\Conversational_AI\Source\data\honda_knowledge_base\Sheet1"  # Replace with the actual folder path
docs = iterate_files(folder_path)
print(docs)
```
